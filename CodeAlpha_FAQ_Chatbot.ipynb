```python
!pip install gradio scikit-learn

import gradio as gr
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# FAQs
faq_questions = [
    "What is AI?",
    "What is Machine Learning?",
    "What is Python?",
    "What is Data Science?",
    "What is ChatGPT?"
]

faq_answers = [
    "Artificial Intelligence is the simulation of human intelligence by machines.",
    "Machine Learning is a subset of AI that allows systems to learn automatically.",
    "Python is a popular programming language.",
    "Data Science is the field of analyzing data to extract insights.",
    "ChatGPT is an AI chatbot developed by OpenAI."
]

# Chatbot Function
def chatbot(user_input):
    all_questions = faq_questions + [user_input]

    vectorizer = CountVectorizer().fit_transform(all_questions)
    vectors = vectorizer.toarray()

    similarity = cosine_similarity([vectors[-1]], vectors[:-1])

    index = np.argmax(similarity)

    return faq_answers[index]

# Gradio Interface
interface = gr.Interface(
    fn=chatbot,
    inputs=gr.Textbox(label="Ask a Question"),
    outputs=gr.Textbox(label="Answer"),
    title="🤖 FAQ Chatbot",
    description="Ask AI-related questions and get answers instantly."
)

interface.launch()
```
